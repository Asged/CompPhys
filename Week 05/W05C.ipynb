```python
xs = np.linspace(-2,2, 100)

def P1(x):
    return 1 / (np.pi * np.sqrt(4-x**2))

def P2(x, kT=2):
    return 1 / np.sqrt(2*np.pi*kT) * np.exp(-x**2 / (2*kT))
```

```python
fig, axes = plt.subplots(1,2,figsize=(12,6))

axes[0].plot(xs, P1(xs))
axes[1].plot(xs, P2(xs))
```

```python
P1_integrated = quad(P1, -2,2)[0]
P2_integrated = quad(P2, -10,10)[0]
print(P1_integrated, P2_integrated)
```
